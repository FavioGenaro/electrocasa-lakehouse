# Event Log

Últimos eventos

```sql
SELECT
    timestamp,
    level,
    event_type,
    message,
    error
FROM electrocasa.monitoring.event_log_electrocasa
ORDER BY timestamp DESC
LIMIT 100;
```

Errores del pipeline

```sql
SELECT
    timestamp,
    event_type,
    message,
    error,
    origin.pipeline_id,
    origin.update_id,
    origin.flow_name
FROM electrocasa.monitoring.event_log_electrocasa
WHERE level = 'ERROR'
ORDER BY timestamp DESC;
```

Monitorear los flujos

```sql
SELECT
    timestamp,
    origin.flow_name AS flow_name,
    origin.update_id AS update_id,
    details:flow_progress.status AS estado,
    message
FROM electrocasa.monitoring.event_log_electrocasa
WHERE event_type = 'flow_progress'
ORDER BY timestamp DESC;
```

Monitorear calidad de los datos recibidos

```sql
SELECT
    timestamp,
    origin.flow_name AS flow_name,
    details:flow_progress.data_quality AS data_quality
FROM electrocasa.monitoring.event_log_electrocasa
WHERE event_type = 'flow_progress'
  AND origin.flow_name LIKE '%.bronze.%'
  AND details:flow_progress.data_quality IS NOT NULL
ORDER BY timestamp DESC;
```
